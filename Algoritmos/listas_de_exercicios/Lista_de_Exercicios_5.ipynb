#### 5. Peça ao usuário uma palavra e conte quantas vezes cada letra aparece. Armazene as contagens em um dicionário e imprima o resultado.

````python
palavra, count = input(), {}

for i in palavra:
    if i in list(count.keys()):
        pass
    else:
        count[i] = 0
        for j in palavra:
            if i == j:
                count[i] += 1

print(count)
````

#### 6. Utilizando um dicionário, armazene os preços de três produtos fornecidos pelo usuário. Depois, mostre todos os produtos e seus preços formatados com duas casas decimais.

````python
produtos = {input("Insira nome do produto: "): round(float(input("Insira preço do produto: ")), 2) for _ in range(3)}
for produto, preco in produtos.items(): print(f"{produto} - {preco:.2f}")
````

#### 7. Crie um dicionário com três nomes de alunos como chave e listas de três notas cada como valor. Em seguida, calcule e imprima a média de cada aluno.

````python
alunos = {input("Insira nome do aluno: "): [float(input("Insira uma nota: ")) for i in range(3) ] for _ in range(3)}
for aluno, notas in alunos.items():
    media = 0
    for nota in notas:
        media += nota
    media /= len(notas)
    print(f"Media do {aluno}: {media}")
````

#### 11. Crie um programa que leia várias palavras digitadas pelo usuário até ele digitar `"fim"`. Para cada palavra digitada, conte quantas vezes ela foi repetida usando um dicionário. Por fim, imprima o resultado.

````python
count = {}

while (temp := input()) != "fim":
    count[temp] = 1 if temp not in count.keys() else count[temp] + 1
print(count)
````

#### 13. Monte um sistema simples de cadastro de alunos, onde você armazena nome e notas em um dicionário. Depois, calcule e imprima a média das notas.

````python
alunos = []

while True:
    try:
        print("[0 - Sair | 1 - Cadastrar | 2 - Exibir Alunos ]")
        choice = input()
        match choice:
            case '0': 
                print("Desligando Sistema...")
                break
            case '1':
                alunos.append({
                    'nome': input("Inserir nome do Aluno: "),
                    'notas': [float(input(f"Inserir {i + 1}° nota: ")) for i in range(4)]
                })
            case '2':
                for aluno in alunos:
                    print(f"Aluno: {aluno['nome']} | Notas: {', '.join(f"{i}" for i in aluno['notas'])} | Media: {sum(aluno['notas']) / len(aluno['notas'])}")
                input()
            case _:
                pass
    except:
        input("3RR0R! Aperte enter para reiniciar.")
        pass
````

#### 14. Dado um texto digitado pelo usuário, conte quantas vezes cada palavra aparece usando um dicionário. Ignore diferenças entre maiúsculas e minúsculas.

````python
count = {}

while (temp := input().lower()) != "fim":
    count[temp] = 1 if temp not in count.keys() else count[temp] + 1
print(count)
````
