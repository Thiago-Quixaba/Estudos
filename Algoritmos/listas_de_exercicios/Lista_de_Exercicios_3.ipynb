#### 2. Escreva um algoritmo que some os elementos de uma lista de números.

````python
lista = []
soma = 0

while True:
    try:
        lista.append(input())
    except:
        break

for i in lista:
    soma += i

print(soma)
````

#### 3. Escreva um algoritmo que encontre o maior valor de uma lista.

````python
lista = []

while True:
    try:
        lista.append(float(input()))
    except:
        break

max = lista[0]

for i in lista:
    if i > max:
        max = i

print(max)
````

#### 4. Escreva um algoritmo que analise uma lista e informe quantos elementos dessa lista são pares.

````python
lista = []

while True:
    try:
        lista.append(float(input()))
    except:
        break

count = 0

for i in lista:
    if i % 2 == 0:
        count += 1

print(count)
````

#### 6. Escreva um algoritmo que remova todos os elementos negativos de uma lista.

````python
lista = []

while True:
    try:
        lista.append(float(input()))
    except:
        break

for i in lista:
    if i < 0:
        lista.remove(i)

print(', '.join(lista))
````

#### 8. Escreva um algoritmo que verifique se uma lista está em ordem crescente.

````python
lista = []
crescente = True

while True:
    try:
        lista.append(float(input()))
    except:
        break

for i in range(len(lista) - 1):
    if lista[i] > lista[i + 1]:
        crescente = False
        break

print(crescente)
````

#### 11. Leia uma lista de 5 posições e, em seguida, um valor X qualquer. O programa deverá fazer uma busca do valor X no vetor lido e informar a posição em que foi encontrado ou se não foi encontrado.

````python
lista, x, retorno = [input() for _ in range(5)], input(), []
if lista.count(x) != 0:
    for i in range(5):
        if x == lista[i]: 
            retorno.append(i)
    print(f"O {x} foi encontrado nas posições: {', '.join([str(_) for _ in retorno])}")
else:
    print(f"O {x} não foi encontrado")
````

#### 12. Leia uma lista com 5 elementos, verifique se existem elementos iguais e os escreva.

````python
lista, retorno = [input() for _ in range(5)], []
for i in lista:
    if i not in retorno:
        if lista.count(i) > 1:
            retorno.append(i)
print(', '.join(retorno))
````

#### 14. Leia uma matriz 3 x 3 e escreva a localização (linha e coluna) do maior valor.

````python
matriz = [[float(input()) for j in range(3)] for i in range(3)]

max = [[0, 0]]
for i in range(3):
    for j in range(3):
        if matriz[max[0][0]][max[0][1]] < matriz[i][j]:
            max = [[i, j]]
        elif matriz[max[0][0]][max[0][1]] == matriz[i][j]:
            max.append([i, j])

print(max)
````

#### 15. Leia duas matrizes 3 x 3 e escreva uma terceira matriz com os maiores elementos entre as duas matrizes.

#### 17. Quadrados Mágicos consistem em uma matriz quadrada em que as somas das linhas, das colunas e das duas diagonais principais são iguais.

**Exemplo (3x3):**
```
6  1  8
7  5  3
2  9  4
```

#### Nesse caso, todas as somas são iguais a 15. Faça um programa que: Leia 9 dígitos, Verifique se não há repetições (devem ser exatamente os números de 1 a 9), Monte o quadrado, Informe se é ou não um quadrado mágico

````python
matriz, quadradoMagico, numeros = [], True, [i for i in range(1, 10)]

for j in range(3):
    matriz.append([])
    for i in range(3):
        aux = int(input())
        if aux not in numeros:
            quadradoMagico = False
        else:
            numeros.remove(aux)
        matriz[j].append(aux)

if quadradoMagico:
    somas = []

    for i in matriz:
        aux = 0
        for j in i:
            aux += j
        somas.append(aux)
    
    for i in range(3):
        aux = 0
        for j in matriz:
            aux += j[i]
        somas.append(aux)
    
    for j in range(2):
        aux = 0
        if j == 0:
            for i in range(3):
                aux += matriz[i][i]
        else:
            for i in range(3):
                aux += matriz[i][2 - i]
        somas.append(aux)
    
    for i in somas:
        if i != somas[0]:
            quadradoMagico = False
            break

print(quadradoMagico)
````
