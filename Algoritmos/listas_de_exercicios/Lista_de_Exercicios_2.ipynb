#### 6. Escreva um algoritmo que calcule o fatorial de um número digitado pelo usuário.
Na matemática, o fatorial de um número natural n, denotado por n!, é o produto de todos os naturais menores ou iguais a n.
**Fórmula:**
n! = n × (n - 1)! × ... × 1

````python
try:
    num = int(input())
    if num < 0:
        print()
    elif num == 1 or num == 0:
        print(1)
    else:
        for _ in range(2, num):
            num *= _
        print(num)
except:
    print()
````

#### 8. Escreva um algoritmo que verifique se um determinado número digitado pelo usuário é primo. Números primos são todos os números naturais que possuem somente dois divisores: 1 e eles mesmos.

````python
prime, num = True, int(input())
if num <= 1:
    prime = False
elif num == 2:
    pass
elif num % 2 == 0:
    prime = False
else:
    for _ in range(3, ((num // 2) + 1), 2):
        if num % _ == 0:
            prime = False
            break
print(prime)
````
