28. Um caixa eletrônico deve entregar o menor número de notas possíveis para um saque. Disponível: notas de 100, 50, 20, 10, 5, 2 e 1.

```python
number = int(input()) # o problema tá muito vago
values = {'100': 0, '50': 0, '20': 0, '10': 0, '5': 0, '2': 0, '1': 0}
while number >= 1:
    if number >= 100:
        number -= 100
        values['100'] += 1
    elif number >= 50:
        number -= 50
        values['50'] += 1
    elif number >= 20:
        number -= 20
        values['20'] += 1
    elif number >= 10:
        number -= 10
        values['10'] += 1
    elif number >= 5:
        number -= 5
        values['5'] += 1
    elif number >= 2:
        number -= 2
        values['2'] += 1
    else:
        number -= 1
        values['1'] += 1
print(values)
```
