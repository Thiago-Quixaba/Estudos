#### 3. Escreva um algoritmo que multiplique  duas matrizes 3x3 e exiba o resultado.

````python
matriz, matriz1, matriz2 = [[0 for j in range(3)] for i in range(3)], [[float(input()) for j in range(3)] for i in range(3)], [[float(input()) for j in range(3)] for i in range(3)]

for i in range(3):
    for l in range(3):
        matriz[i][l] = matriz1[i][l] * matriz2[i][l]

print(matriz)
````

#### 4. Escreva um algoritmo que calcule a soma de todos os elementos da matriz.

#### 5. Escreva um algoritmo que imprima a diagonal principal e a secundária.

````python
matriz = [[input() for j in range(3)] for i in range(3)]

print("Diagonal Principal:", end=' ')
for i in range(len(matriz)):
    print(matriz[i][i], end=' ')

print()

print("Diagonal Secundaria:", end=' ')
for i in range(len(matriz)):
    print(matriz[i][(len(matriz[i]) - 1) - i], end=' ')
````

#### 6. Escreva um algoritmo que substitua todos os números pares da matriz por 0.

````python
matriz = [[float(input()) for j in range(3)] for i in range(3)]

for i in range(len(matriz)):
    for j in range(len(matriz[i])):
        if matriz[i][j] % 2 == 0:
            matriz[i][j] = 0

print(matriz)
````

#### 7. Crie um algoritmo que percorra uma matriz 4x4 e:
- a. encontre o maior valor
- b. encontre o menor valor
- c. informe suas posições (linha e coluna)

````python
matriz = [[float(input()) for j in range(4)] for i in range(4)]
max_val, min_val = [matriz[0][0], []], [matriz[0][0], []]

for i in [0, 1, 2, 3]:
    for j in [0, 1, 2, 3]:
        if max_val[0] == matriz[i][j]:
            max_val[1].append([i, j])
        elif max_val[0] < matriz[i][j]:
            max_val = [matriz[i][j], [[i, j]]]
        
        if min_val[0] == matriz[i][j]:
            min_val[1].append([i, j])
        elif min_val[0] > matriz[i][j]:
            min_val = [matriz[i][j], [[i, j]]]

print(f"Maior valor: {max_val[0]} - Posições: {'; '.join([f"{i}, {j}" for i, j in max_val[1]])}\nMenor valor: {min_val[0]} - Posições: {'; '.join([f"{i}, {j}" for i, j in min_val[1]])}")
````

#### 8. Dada uma matriz 5x5:
- a. calcule a soma de cada linha
- b. calcule a soma de cada coluna
- c. exiba os resultados separadamente

````python
matriz, soma_linhas, soma_colunas = [[float(input()) for j in range(5)] for i in range(5)], [], []

for i in range(5):
    soma_linhas.append(0)
    soma_colunas.append(0)
    for j in range(5):
        soma_linhas[i] += matriz[i][j]
        soma_colunas[i] += matriz[j][i]
    soma_linhas[i] = str(soma_linhas[i])
    soma_colunas[i] = str(soma_colunas[i])

print(f"Linhas: {', '.join(soma_linhas)}\nColunas: {', '.join(soma_colunas)}")
````

#### 9. Faça um algoritmo que:
- a. conte quantos números pares existem na matriz
- b. conte quantos números ímpares existem

````python
matriz, pares, impares = [[float(input()) for j in range(3)] for i in range(3)], 0, 0

for i in matriz:
    for j in i:
        if j % 2 == 0:
            pares += 1
        else:
            impares += 1


print(f"Pares: {pares}\nImpares: {impares}")
````

#### 10. Uma matriz é simétrica quando:
```
𝐴[𝑖][𝑗] = 𝐴[𝑗][𝑖]
```
#### Crie um algoritmo que:
- a. verifique se uma matriz 3x3 é simétrica
- b. informe o resultado ao usuário

````python
matriz, simetrica = [[float(input()) for j in range(3)] for i in range(3)], True

for i in range(3):
    if simetrica:
        for j in range(3):
            if matriz[i][j] != matriz[j][i]:
                simetrica = False
                break
    else:
        break

print(simetrica)
````

#### 11. Crie um algoritmo que:
- a. peça um número ao usuário
- b. verifique se ele existe na matriz
- c. informe todas as posições onde ele aparece

````python
matriz, x, loc = [[float(input()) for j in range(3)] for i in range(3)], float(input()), []

for i in range(len(matriz)):
    for j in range(len(matriz[i])):
        if x == matriz[i][j]:
            loc.append([i, j])


print("Valor não encontrado" if not loc else loc)
````

#### 13. Crie um algoritmo que some apenas os elementos da borda de uma matriz 3x3.

````python
matriz, soma = [[float(input()) for j in range(3)] for i in range(3)], 0

for i in range(len(matriz)):
    for j in range(len(matriz[i])):
        if (i == 0 or i == len(matriz) - 1) or (j == 0 or j == len(matriz[i]) - 1):
            soma += matriz[i][j]

print(soma)
````

#### 14. Uma matriz é triangular superior quando todos os elementos abaixo da diagonal principal são zero. Crie um algoritmo que Verifique se a matriz atende essa condição.

````python
matriz, triangular = [[float(input()) for j in range(3)] for i in range(3)], True

for i in range(len(matriz)):
    for j in range(len(matriz[i])):
        triangular = False if i > j and matriz[i][j] != 0 else triangular

print(triangular)
````
