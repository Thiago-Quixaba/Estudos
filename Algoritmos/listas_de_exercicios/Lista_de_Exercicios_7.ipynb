1. Escreva uma função que receba uma lista de números e retorne uma nova lista com apenas os números primos.

```python
def verify_prime(lista: list[int] = []) -> list[int]:
    lista_primos = lista.copy()
    for num in lista:
        if num <= 1:
            lista_primos.remove(num)
            continue

        if num == 2:
            continue
        elif num % 2 == 0:
            lista_primos.remove(num)
            continue

        for _ in range(3, ((num // 2) + 1), 2):
            if num % _ == 0:
                lista_primos.remove(num)
                break
        
    return lista_primos

print(verify_prime([]))
```

6. Escreva uma função que conte quantas vezes cada palavra aparece em uma frase.

```python
def count(frase: str = "") -> dict:    
    palavras = {}

    for palavra in frase.strip().split(' '):
        if palavra == '':
            continue
        if palavra not in palavras:
            palavras[palavra] = 1
        else:
            palavras[palavra] += 1
    
    return palavras if palavras != {} else None

print(count(""))
```

7. Crie uma função que receba uma lista de números e devolva um dicionário com estatísticas:
- a. soma
- b. média
- c. valor mínimo
- d. valor máximo

```python
def nao_sei(lista: list[int] = []) -> list[int]:
    lista.sort()

        
    return {
        'soma': sum(lista),
        'media': sum(lista) / len(lista),
        'valor_mínimo': lista[0],
        'valor_máximo': lista[-1]
    }

print(nao_sei())
```

10. Crie uma função que calcule o MDC (máximo divisor comum) entre dois números, utilizando o algoritmo de Euclides.

```python
def MDC_Euclides(num1: int = 1, num2: int = 1):
    if num1 < num2:
        num1, num2 = num2, num1
    
    resto = num1 % num2

    if resto == 0:
        return num2
    else:
        return MDC_Euclides(num2, resto)

print(MDC_Euclides(48, 18))
```

11. Escreva uma função que verifique se duas palavras são anagramas (mesmas letras em ordem diferente). Ex: amor – Roma = True; casa – caro = False

```python
def count(frase: str = "") -> dict:    
    letras = {}

    for letra in list(frase.strip()):
        if letra == '':
            continue
        if letra not in letras:
            letras[letra] = 1
        else:
            letras[letra] += 1
    
    return letras if letras != {} else None

verify_anagrama = lambda palavra1, palavra2: count(palavra1) == count(palavra2)

print(verify_anagrama())
```

13. Crie uma função que receba uma lista e retorne uma nova lista com os elementos sem repetição, mantendo a ordem original.

```python
def nao_sei(lista: list = []):
    nova_lista = []
    for item in lista:
        if item not in nova_lista:
            nova_lista.append(item)

    return nova_lista

print(nao_sei([1, 1, 1, 2, 2, 3, 4, 3, 5]))
```

14. Crie uma função recursiva que calcule a soma dos números de 1 até n.

16. Crie uma função recursiva que imprima uma contagem regressiva de n até 1.

```python
def recursiva(n: int = 1):
    if n != 0:
        print(n)
        recursiva(n - 1)

recursiva(10)
```
