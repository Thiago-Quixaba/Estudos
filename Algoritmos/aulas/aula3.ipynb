```python
senha, count = "1234", 0
while True:
    count += 1
    senha_digitada = input("Digite sua senha: ")
    if senha_digitada == senha:
        print("Acesso autorizado!")
        break
    elif count >= 3:
        print("Acesso negado!")
        break
    else:
        print("Senha incorreta")
```

```python

from random import randrange
numero_sorteado = randrange(10)
tentativas = 0
print("Você tem 3 tentativas para acertar o número sorteado.")
while tentativas < 3:
    numero_digitado = int(input("Digite um número: "))
    if numero_digitado == numero_sorteado:
        print("Parabéns, você acertou!")
        break
    else:
        print(f"Você errou! O número correto é mais {"alto" if numero_digitado < numero_sorteado else "baixo"}! Tente novamente.")
```
