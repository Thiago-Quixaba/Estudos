```python
numero = int(input("Digite um numero: "))
if numero > 10:
    print("O numero digitado é maior que 10")
elif numero == 10:
    print("O numero digitado é igual a 10")
else:
    print("O numero digitado é menor que 10")
```

Verifique se o numero digitado pelo usuario é par ou impar e imprima uma mensagem em cada caso

```python
print("O numero digitado é par" if int(input("Digite um numero: ")) % 2 == 0 else "O numero digitado é impar")
```
