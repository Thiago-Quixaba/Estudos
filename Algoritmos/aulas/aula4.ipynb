```python
A, B, soma = [[1, 2, 3], [4, 5, 6], [7, 8, 9,]], [[9, 8, 7], [6, 5, 4], [3, 2, 1]], 0
C = [[(A[i][0] * B[0][j] + A[i][1] * B[1][j] + A[i][2] * B[2][j]) for j in range(3)] for i in range(3)]
min_value, max_value = [[C[0][0], []] for i in range(2)]
for i in range(3):
    soma += C[i][i]
    for j in range(3):
        if C[i][j] == max_value[0]:
            max_value[1].append(f'{i}, {j}')
        elif C[i][j] > max_value[0]:
            max_value = [C[i][j], [f'{i}, {j}']]
        if C[i][j] == min_value[0]:
            min_value[1].append(f'{i}, {j}')
        elif C[i][j] < min_value[0]:
            min_value = [C[i][j], [f'{i}, {j}']]
print(f"{C} - Soma da Diagonal Principal: {soma}\nMaior valor: {max_value[0]} - Posições: {'; '.join(max_value[1])}\nMenor valor: {min_value[0]} - Posições: {'; '.join(min_value[1])}")
```
